## Parte 2: encuentra los errores (6 pts)

Un practicante valoró a la empresa Andina SAC y entregó este trabajo. Sus datos de partida son correctos: EBIT = 50, t = 30 por ciento, depreciación = 10, capex = 15, incremento del capital de trabajo = 5, gasto por intereses = 8, endeudamiento neto = 0, Ke = 12 por ciento, WACC = 9 por ciento.

> **Paso 1.** FCFF = EBIT(1-t) + Dep + Int - Capex - WCInv = 35 + 10 + 8 - 15 - 5 = **33**
>
> **Paso 2.** FCFE = FCFF - Int = 33 - 8 = **25**
>
> **Paso 3.** Como la empresa creció 7 por ciento en los últimos años, uso g = 7 por ciento a perpetuidad. Valor del equity = FCFE por (1+g) entre (WACC - g) = 25 por 1.07 entre 0.02 = **1,337.5**

El trabajo contiene **cuatro errores conceptuales**. Identifícalos (2 pts), explica en una oración por qué cada uno es un error (2 pts), y recalcula el valor del equity corrigiendo todo, con un g perpetuo de 3 por ciento (2 pts). Puedes usar la celda de código.

*Errores y explicaciones:*

1. En el paso 1, el EBIT significa la utilidad operativa antes de intereses por lo que adicionar nuevamente los intereses estaría sobreestimando
2. En el paso 2, hace falta descontar el interés después de impuesto Int(1-t) porque la empresa podría utilizar el escudo tributario de tener deuda. 
3. Como lo hablamos en la última clase, en el paso 3, tener un crecimiento de 7% a perpetuidad no es sostenible en el largo plazo
4. En el  último paso el error es que por regla de consistencia el FCFE se descuenta con Ke en lugar de WACC.

````python
# Recalculo corregido (puedes usar utils/finanzas.py si trabajas en tu fork)

# Mi código
EBIT = 50
t = 0.30
dep = 10
capex = 15
wc_inv = 5
intereses = 8
endeudamiento_neto = 0

Ke = 0.12
g = 0.03

FCFF = EBIT * (1 - t) + dep - capex - wc_inv
FCFE = FCFF - intereses * (1 - t) + endeudamiento_neto
valor_equity = FCFE * (1 + g) / (Ke - g)

print(f"FCFF corregido: {FCFF:.2f}")
print(f"FCFE corregido: {FCFE:.2f}")
print(f"Valor del Equity: {valor_equity:.2f}")
````

## Parte 3: tu empresa (6 pts)

Con la empresa de tus tareas (usa `yfinance` o los resultados que ya tienes de la tarea 2):

**3.1** Calcula el FCFF del último año disponible por la ruta del EBIT. Muestra los insumos que usaste (EBIT, tasa efectiva, depreciación, capex, capital de trabajo). (3 pts)

````python
# TU CODIGO AQUI: FCFF del ultimo anio de tu empresa, ruta EBIT
import yfinance as yf
import pandas as pd

# Empresa
ticker = yf.Ticker("KO")

# Estados financieros anuales
income = ticker.financials
cashflow = ticker.cashflow
balance = ticker.balance_sheet

# Último año disponible
ultimo_anio = income.columns[0]

print("Último año disponible:", ultimo_anio)
print("\nEBIT:")
print(income.loc["EBIT", ultimo_anio])

print("\nDepreciación y amortización:")
print(cashflow.loc["Depreciation And Amortization", ultimo_anio])

print("\nCAPEX:")
print(cashflow.loc["Capital Expenditure", ultimo_anio])

# Tasa efectiva de impuestos

pretax_income = income.loc["Pretax Income", ultimo_anio]
tax_provision = income.loc["Tax Provision", ultimo_anio]

tasa_efectiva = tax_provision / pretax_income

print(f"Utilidad antes de impuestos: ${pretax_income:,.0f}")
print(f"Impuesto: ${tax_provision:,.0f}")
print(f"Tasa efectiva: {tasa_efectiva:.2%}")

# Capital de trabajo

current_assets = balance.loc["Current Assets", ultimo_anio]
current_liabilities = balance.loc["Current Liabilities", ultimo_anio]

anio_anterior = income.columns[1]

current_assets_ant = balance.loc["Current Assets", anio_anterior]
current_liabilities_ant = balance.loc["Current Liabilities", anio_anterior]

capital_trabajo = current_assets - current_liabilities
capital_trabajo_ant = current_assets_ant - current_liabilities_ant

delta_wc = capital_trabajo - capital_trabajo_ant

print(f"Capital de trabajo último año: ${capital_trabajo:,.0f}")
print(f"Capital de trabajo año anterior: ${capital_trabajo_ant:,.0f}")
print(f"Δ Capital de trabajo: ${delta_wc:,.0f}")

# Cálculo del FCFF por la ruta del EBIT

EBIT = income.loc["EBIT", ultimo_anio]
depreciacion = cashflow.loc["Depreciation And Amortization", ultimo_anio]
capex = abs(cashflow.loc["Capital Expenditure", ultimo_anio])

FCFF = EBIT * (1 - tasa_efectiva) + depreciacion - capex - delta_wc

print("----- FCFF DE COCA-COLA -----")
print(f"EBIT: ${EBIT:,.0f}")
print(f"Tasa efectiva: {tasa_efectiva:.2%}")
print(f"Depreciación y amortización: ${depreciacion:,.0f}")
print(f"CAPEX: ${capex:,.0f}")
print(f"Δ Capital de trabajo: ${delta_wc:,.0f}")
print(f"\nFCFF: ${FCFF:,.0f}")
````
